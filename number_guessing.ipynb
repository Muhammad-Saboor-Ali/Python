```python
import random

def play_game():
    
    print("\nSelect Difficulty Level:")
    print("1. Easy (1-50)")
    print("2. Medium (1-100)")
    print("3. Hard (1-200)")
    
    choice = input("Enter choice (1/2/3): ")
    
    if choice == '1':
        max_range = 50
    elif choice == '2':
        max_range = 100
    else:
        max_range = 200
    
    number = random.randint(1, max_range)
    attempts = 0
    
    print(f"\nI am thinking of a number between 1 and {max_range}")
    
    while True:
        try:
            guess = int(input("Enter your guess: "))
            attempts += 1
            
            if guess > number:
                print(" Too High!")
            elif guess < number:
                print(" Too Low!")
            else:
                print("🎉 Correct!")
                print("You guessed in", attempts, "attempts.")
                break
        
        except ValueError:
            print(" Please enter a valid number!")

while True:
    play_game()
    again = input("\nDo you want to play again? (yes/no): ")
    if again.lower() != "yes":
        print("Thanks for playing!")
        break
```
